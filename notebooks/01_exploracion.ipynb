```python
# ===== CARGA =====
import pandas as pd

URL_BASE = "https://raw.githubusercontent.com/zygmuntz/goodbooks-10k/master/"
books = pd.read_csv(URL_BASE + "books.csv")
```

```python
print(books.shape)
books.head()
```

```python
books.info()
```

```python
# ===== CLAVES =====
for columna in ["book_id", "goodreads_book_id", "best_book_id", "work_id"]:
    print(columna, books[columna].nunique())
```

```python
# ===== VALORACIONES =====
ratings = pd.read_csv(URL_BASE + "ratings.csv")
print(ratings.shape)
ratings.head()
```

```python
print(ratings.isnull().sum())
print("Usuarios:", ratings["user_id"].nunique())
print("Libros:", ratings["book_id"].nunique())
ratings["rating"].value_counts().sort_index()
```
